# Data Cleaning: Willow & Wick 2025 Sales
Cleaning the client's raw sales export before analysis.

## 1. Explore the raw data

```python
import pandas as pd

df = pd.read_csv("../data/raw/willow_and_wick_sales_2025.csv")
df.head(10)
```

```python
print(df.shape)
df.info()
```

# 2. Data quality problems found

```python
print("Fully duplicated rows:", df.duplicated().sum())
print("Repeated order IDs:", df["order_id"].duplicated().sum())
```

```python
df["quantity"].describe()
```

```python
returns = df[df["quantity"] < 0]
print("Number of return rows:", len(returns))
returns.head()
```

## 3. Remove duplicates

```python
clean = df.copy()

before = len(clean)
clean = clean.drop_duplicates()
after = len(clean)

print(f"Removed {before - after} duplicates rows, Rows now: {after}")
```

## 4. Fix prices and dates

```python
clean["unit_price"] = (
    clean["unit_price"]
    .str.replace("£", "", regex=False)
    .astype(float)
)

print(clean["unit_price"].dtype)
clean["unit_price"].describe()
```

```python
s = clean["order_date"]

dates = pd.to_datetime(s, format="%Y-%m-%d", errors="coerce")
dates = dates.fillna(pd.to_datetime(s, format="%d/%m/%Y", errors="coerce"))
dates = dates.fillna(pd.to_datetime(s, format="%d %b %Y", errors="coerce"))

clean["order_date"] = dates

print("Unparsed dates:", clean["order_date"].isna().sum())
print("Range:", clean["order_date"].min(), "to", clean["order_date"].max())
clean[["order_id", "order_date"]].head(8)
```

```python
print("dates variable type:", dates.dtype)
print("clean column type:", clean["order_date"].dtype)
```

```python
clean.columns.tolist()
```

## 5. Standardise product names

```python
print("Different product names:", clean["product"].nunique())
clean["product"].value_counts()
```

```python
clean["product"] = clean["product"].str.strip().str.title()
print("Different product names:", clean["product"].nunique())
clean["product"].value_counts()
```

## 6. Clean regions

```python
print("Different region:", clean["region"].nunique())
clean["region"].value_counts(dropna=False)
```

## 7. Fill missing values

```python
clean["region"] = clean["region"].str.strip().str.title()
clean["region"] = clean["region"].replace({
    "Se": "South East",
    "South-East": "South East",
})

clean["region"] = clean["region"].fillna("Unknown")

print("Different regions:", clean["region"].nunique())
clean["region"].value_counts()
```

```python
known = clean.dropna(subset=["category"])
print("Max categories per product:", known.groupby("product")["category"].nunique().max())
product_to_category = known.groupby("product")["category"].first()
clean["category"] = clean["category"].fillna(clean["product"].map(product_to_category))
print("Blank categories left:", clean["category"].isna().sum())
clean["category"].value_counts()
```

```python
clean["customer_id"] = clean["customer_id"].fillna("Guest")
print("Blank customer IDs left:", clean["customer_id"].isna().sum())
print("Orders with Guest customer:", (clean["customer_id"] == "Guest").sum())
```

## 8. Flag returns and calculate revenue

```python
clean["is_return"] = clean["quantity"] < 0
clean["revenue"] = clean["quantity"] * clean["unit_price"]

print("Return rows:", clean["is_return"].sum())
print("Gross sales:   £", round(clean.loc[~clean["is_return"], "revenue"].sum(), 2))
print("Returns:       £", round(clean.loc[clean["is_return"], "revenue"].sum(), 2))
print("Net revenue:   £", round(clean["revenue"].sum(),2))
```

```python
discount_rate = clean["discount_code"].map({"10%": 0.10, "15%": 0.15}).fillna(0)
clean["net_revenue"] = (clean["revenue"] * (1 - discount_rate)).round(2)

print("Discounts given: £", round((clean["revenue"] - clean["net_revenue"]).sum(), 2))
print("Net revenue after discounts & returns: £", round(clean["net_revenue"].sum(),2))
```

## 9. Save the clean dataset

```python
from pathlib import Path
Path("../data/processed").mkdir(exist_ok=True)
clean.to_csv("../data/processed/willow_and_wick_sales_2025_clean.csv",index=False)
print("Saved:", len(clean), "rows,", len(clean.columns), "columns")
```
