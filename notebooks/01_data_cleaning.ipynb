```python
import pandas as pd

df = pd.read_csv("../data/raw/willow_and_wick_sales_2025.csv")
df.head(10)
```

```python
print(df.shape)
df.info()
```

```python
print("Fully duplicated rows:", df.duplicated().sum())
print("Repeated order IDs:", df["order_id"].duplicated().sum())
```

```python
df["quantity"].describe()
```

```python
returns = df[df["quantity"] < 0]
print("Number of return rows:", len(returns))
returns.head()
```

```python
clean = df.copy()

before = len(clean)
clean = clean.drop_duplicates()
after = len(clean)

print(f"Removed {before - after} duplicates rows, Rows now: {after}")
```

```python
clean["unit_price"] = (
    clean["unit_price"]
    .str.replace("£", "", regex=False)
    .astype(float)
)

print(clean["unit_price"].dtype)
clean["unit_price"].describe()
```

```python
s = clean["order_date"]

dates = pd.to_datetime(s, format="%Y-%m-%d", errors="coerce")
dates = dates.fillna(pd.to_datetime(s, format="%d/%m/%Y", errors="coerce"))
dates = dates.fillna(pd.to_datetime(s, format="%d %b %Y", errors="coerce"))

clean["order_date"] = dates

print("Unparsed dates:", clean["order_date"].isna().sum())
print("Range:", clean["order_date"].min(), "to", clean["order_date"].max())
clean[["order_id", "order_date"]].head(8)
```

```python
print("dates variable type:", dates.dtype)
print("clean column type:", clean["order_date"].dtype)
```
